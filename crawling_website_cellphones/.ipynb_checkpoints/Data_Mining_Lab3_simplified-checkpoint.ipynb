# Data Mining Lab 3: CellphoneS Data Preparation
Workflow: load raw data -> check quality -> remove duplicates -> extract features from product names -> summarize -> first figure.

## 1. Setup
Import libraries, set the data path and figure folder, and create `RES`, a dictionary that stores key numbers for the report.

```python
%matplotlib inline
import os
import re
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

warnings.filterwarnings('ignore')
sns.set_theme(style='whitegrid', font_scale=0.95)

DATA_PATH = 'cellphones_clean.csv'
FIG_DIR = 'figures'
os.makedirs(FIG_DIR, exist_ok=True)

PAL = {
    'Apple': '#4C4C4C',
    'Samsung': '#1F77B4',
    'Chinese brands': '#D62728',
    'Vietnamese brands': '#2CA02C',
    'Other international': '#FF7F0E',
}
M = 1e6    # display prices in million VND
RES = {}   # key numbers collected along the way


def savefig(name):
    """Save the current figure to FIG_DIR and show it in the notebook."""
    plt.tight_layout()
    plt.savefig(f'{FIG_DIR}/{name}.png', dpi=200, bbox_inches='tight')
    plt.show()
    plt.close()
```

## 2. Load raw data and check quality
Load the crawled data, then count duplicates and missing values.

```python
raw = pd.read_csv(DATA_PATH)

RES['n_raw'], RES['n_cols_raw'] = raw.shape
RES['dup_rows'] = int(raw.duplicated().sum())
RES['dup_url']  = int(raw['url'].duplicated().sum())
RES['dup_name'] = int(raw['name'].duplicated().sum())

# listings that share the same product name
dup_names = raw[raw['name'].duplicated(keep=False)][['name', 'url', 'is_contact']]

# missing values per column
missing_raw = pd.DataFrame({
    'missing': raw.isna().sum(),
    'pct': (raw.isna().mean() * 100).round(1),
})
RES['missing_raw'] = missing_raw.to_dict('index')
missing_raw
```

## 3. Consistency checks
Verify that the price and discount columns make sense:
- the discount recomputed from prices matches the `discount` column
- sale price is never higher than the original price
- no zero or negative prices
- "Contact" listings are exactly the listings without a price

```python
chk = raw.dropna(subset=['price_original']).copy()
chk['calc_disc'] = (1 - chk['price_sale'] / chk['price_original']) * 100

RES['max_disc_err']       = float((chk['calc_disc'] - chk['discount']).abs().max())
RES['sale_gt_orig']       = int((chk['price_sale'] > chk['price_original']).sum())
RES['nonpositive_price']  = int((raw['price_sale'] <= 0).sum())
RES['priced_no_orig']     = int((raw['price_sale'].notna() & raw['price_original'].isna()).sum())
RES['contact_eq_noprice'] = bool((raw['is_contact'] == raw['price_sale'].isna()).all())
```

## 4. Remove duplicates
Duplicated product names are "Contact" listings with no price, so keep only the first one.

```python
df = raw.drop_duplicates(subset='name', keep='first').reset_index(drop=True)
```

## 5. Helper functions
Product specs are hidden inside the product name, so we extract them with text rules:
- `brand_of`: brand name (with aliases, e.g. Redmi/Poco -> Xiaomi)
- `mem`: (RAM, storage) in GB
- `conn`: connectivity type (5G / 4G / Wifi)

`GROUP` maps each brand to a broader group; any brand not listed is treated as a Chinese brand.

```python
PREFIX = r'^(máy tính bảng|máy đọc sách|điện thoại trẻ em|điện thoại)\s+'

BRAND_ALIAS = {
    'iphone': 'Apple', 'ipad': 'Apple',
    'redmi': 'Xiaomi', 'poco': 'Xiaomi', 'mi': 'Xiaomi', 'xiaomii': 'Xiaomi', 'black': 'Xiaomi',
    'xperia': 'Sony', 'iqoo': 'Vivo', 'viettel': 'Viettel', 'leitz': 'Sharp', 'red': 'Nubia',
    'moto': 'Motorola', 'zenfone': 'ASUS', 'rog': 'ASUS', 'asus': 'ASUS', 'camon': 'Tecno',
    'matepad': 'Huawei', 'tcl': 'TCL', 'htc': 'HTC', 'zte': 'ZTE', 'oneplus': 'OnePlus',
    'oppo': 'OPPO', 'vivo': 'Vivo', 'honor': 'Honor', 'tecno': 'Tecno', 'realme': 'Realme',
    'nubia': 'Nubia', 'masstel': 'Masstel',
}

GROUP = {
    'Apple': 'Apple', 'Samsung': 'Samsung',
    **{b: 'Vietnamese brands' for b in ['Masstel', 'Viettel', 'Vsmart', 'Mobell', 'Bphone', 'Benco']},
    **{b: 'Other international' for b in ['Google', 'Nokia', 'Sony', 'Motorola', 'Nothing',
                                          'ASUS', 'HTC', 'Sharp', 'Amazon', 'Acer']},
}


def brand_of(name):
    lower = name.lower()
    if 'kindle' in lower:
        return 'Amazon'
    if 'boox' in lower or 'onyx' in lower:
        return 'Onyx Boox'
    first = re.sub(PREFIX, '', name.strip(), flags=re.I).split()[0]
    return BRAND_ALIAS.get(first.lower(), first.capitalize())


def mem(name):
    # return (ram_gb, storage_gb) parsed from the product name
    text = name.replace('|', ' ')
    # pattern "8GB/256GB" -> both RAM and storage
    m = re.search(r'(\d+)\s*GB\s*[\s/\-\(\)]*\s*(\d+)\s*(GB|TB)', text, re.I)
    if m:
        storage = int(m.group(2)) * (1024 if m.group(3).upper() == 'TB' else 1)
        return int(m.group(1)), storage
    # only one number -> >=32 GB is storage, otherwise RAM
    m = re.search(r'(\d+)\s*(GB|TB)', text, re.I)
    if m:
        v = int(m.group(1)) * (1024 if m.group(2).upper() == 'TB' else 1)
        return (np.nan, v) if v >= 32 else (v, np.nan)
    return np.nan, np.nan


def conn(name):
    lower = name.lower()
    if '5g' in lower:
        return '5G'
    if 'wifi' in lower or 'wi-fi' in lower:
        return 'Wifi'
    if '4g' in lower:
        return '4G'
    return 'Unspecified'
```

## 6. Feature engineering
Create new columns from the product name:
- **Brand and brand group**
- **Product type** (Phone / Tablet / E-reader)
- **RAM and storage** (for e-readers, a single GB figure means storage, not RAM)
- **Flags**: connectivity, official product, bundle, model year, discount shown, availability
- **Tiers**: storage and RAM bins

```python
# brand
df['brand'] = df['name'].apply(brand_of)
df['brand_group'] = df['brand'].map(GROUP).fillna('Chinese brands')

# product type
df['product_type'] = np.where(
    df['name'].str.contains('Máy đọc sách', case=False), 'E-reader',
    np.where(df['category'] == 'tablet', 'Tablet', 'Phone'),
)

# RAM / storage
pairs = df['name'].apply(mem)
df['ram_gb'] = [p[0] for p in pairs]
df['storage_gb'] = [p[1] for p in pairs]

# e-readers: the single GB number is storage, so move it and clear RAM
is_ereader = df['product_type'] == 'E-reader'
only_one_number = is_ereader & df['ram_gb'].notna() & df['storage_gb'].isna()
df.loc[only_one_number, 'storage_gb'] = df.loc[only_one_number, 'ram_gb']
df.loc[is_ereader, 'ram_gb'] = np.nan

# other features from the name
df['connectivity'] = df['name'].apply(conn)
df['is_official']  = np.where(df['name'].str.contains('Chính hãng', case=False), 'Official', 'Not stated')
df['has_bundle']   = np.where(df['name'].str.contains('Kèm', case=False), 'Bundle', 'No bundle')
df['model_year']   = df['name'].str.extract(r'\b(20[12]\d)\b')[0].astype(float)
df['has_discount'] = np.where(df['discount'].notna(), 'Discount shown', 'No discount shown')
df['availability'] = np.where(df['is_contact'], 'Contact (no price)', 'Priced')

# tiers (bins)
df['storage_tier'] = (pd.cut(df['storage_gb'], [0, 64, 128, 256, 1e9],
                             labels=['<=64 GB', '128 GB', '256 GB', '>=512 GB'])
                      .astype(object).fillna('Unknown'))
df['ram_tier'] = (pd.cut(df['ram_gb'], [0, 4, 8, 1e9],
                         labels=['<=4 GB', '6-8 GB', '>=12 GB'])
                  .astype(object).fillna('Unknown'))
```

## 7. Summary of the cleaned dataset
`P` is the subset of listings that have a price; it is used for all price analysis later.

```python
N = len(df)
P = df[df['price_sale'].notna()].copy()

RES['N'] = N
RES['n_priced'] = len(P)
RES['n_contact'] = int(df['is_contact'].sum())
RES['derived_cov'] = {c: int(df[c].notna().sum()) for c in ['ram_gb', 'storage_gb', 'model_year']}
RES['n_brands'] = df['brand'].nunique()
RES['contact_by_cat'] = pd.crosstab(df['category'], df['is_contact']).to_dict('index')

# counts of each categorical column (all listings)
for key, col in [('brand_counts', 'brand'), ('group_counts', 'brand_group'),
                 ('type_counts', 'product_type'), ('cat_counts', 'category'),
                 ('conn_counts', 'connectivity')]:
    RES[key] = df[col].value_counts().to_dict()

# counts among priced listings only
RES['priced_by_group'] = P['brand_group'].value_counts().to_dict()
RES['priced_by_type']  = P['product_type'].value_counts().to_dict()
```

## 8. Figure 1: Dataset overview
Stacked bar chart of the top 12 brands by number of listings, split into priced listings and "Contact" listings (no price shown).

```python
top = df['brand'].value_counts().head(12).index
sub = df[df['brand'].isin(top)]
overview = pd.crosstab(sub['brand'], sub['availability']).loc[top]

overview.plot(kind='barh', stacked=True, figsize=(7, 4.2), color=['#BDBDBD', '#1F77B4'])
plt.gca().invert_yaxis()
plt.xlabel('Number of product listings')
plt.ylabel('Brand')
plt.title('Figure 1. Listings per brand (top 12): priced vs "Contact" (no price)')
plt.legend(title='Availability')
savefig('fig0_overview')
```
