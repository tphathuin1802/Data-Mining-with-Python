# Data Mining Lab 3 (Sections 3.1-3.6) - CellphoneS product dataset
Understand -> Describe -> Visualize -> Compare -> Prepare for Classification -> Generalize & Summarize

```python
import os, re, json, warnings
import numpy as np, pandas as pd
import matplotlib; matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
warnings.filterwarnings('ignore')

DATA_PATH = os.environ.get('DATA_PATH', 'cellphones_clean.csv')
FIG_DIR = os.environ.get('FIG_DIR', 'figures'); os.makedirs(FIG_DIR, exist_ok=True)
sns.set_theme(style='whitegrid', font_scale=0.95)
PAL = {'Apple':'#4C4C4C','Samsung':'#1F77B4','Chinese brands':'#D62728','Vietnamese brands':'#2CA02C','Other international':'#FF7F0E'}
M = 1e6  # display prices in million VND
RES = {}

def savefig(name):
    plt.tight_layout(); plt.savefig(f'{FIG_DIR}/{name}.png', dpi=200, bbox_inches='tight'); plt.close()
```

```python
raw = pd.read_csv(DATA_PATH)
RES['n_raw'] = len(raw); RES['n_cols_raw'] = raw.shape[1]
RES['dup_rows'] = int(raw.duplicated().sum()); RES['dup_url'] = int(raw.url.duplicated().sum())
RES['dup_name'] = int(raw.name.duplicated().sum())
dup_names = raw[raw.name.duplicated(keep=False)][['name', 'url', 'is_contact']]
missing_raw = pd.DataFrame({'missing': raw.isna().sum(), 'pct': (raw.isna().mean()*100).round(1)})
RES['missing_raw'] = missing_raw.to_dict('index')
# consistency checks
chk = raw.dropna(subset=['price_original']).copy()
chk['calc_disc'] = (1 - chk.price_sale/chk.price_original)*100
RES['max_disc_err'] = float((chk.calc_disc - chk.discount).abs().max())
RES['sale_gt_orig'] = int((chk.price_sale > chk.price_original).sum())
RES['nonpositive_price'] = int((raw.price_sale <= 0).sum())
RES['priced_no_orig'] = int((raw.price_sale.notna() & raw.price_original.isna()).sum())
RES['contact_eq_noprice'] = bool((raw.is_contact == raw.price_sale.isna()).all())

# drop the duplicated listing (same product name, both "contact" listings) -> keep first
df = raw.drop_duplicates(subset='name', keep='first').reset_index(drop=True)

def brand_of(n):
    l = n.lower()
    if 'kindle' in l: return 'Amazon'
    if 'boox' in l or 'onyx' in l: return 'Onyx Boox'
    s = re.sub(r'^(máy tính bảng|máy đọc sách|điện thoại trẻ em|điện thoại)\s+', '', n.strip(), flags=re.I)
    t = s.split()[0].lower()
    m = {'iphone':'Apple','ipad':'Apple','redmi':'Xiaomi','poco':'Xiaomi','mi':'Xiaomi','xiaomii':'Xiaomi','black':'Xiaomi',
         'xperia':'Sony','iqoo':'Vivo','viettel':'Viettel','leitz':'Sharp','red':'Nubia','moto':'Motorola','zenfone':'ASUS',
         'rog':'ASUS','camon':'Tecno','matepad':'Huawei','tcl':'TCL','htc':'HTC','zte':'ZTE','asus':'ASUS','oneplus':'OnePlus',
         'oppo':'OPPO','vivo':'Vivo','honor':'Honor','tecno':'Tecno','realme':'Realme','nubia':'Nubia','masstel':'Masstel'}
    return m.get(t, s.split()[0].capitalize())

def mem(n):
    n2 = n.replace('|', ' ')
    m = re.search(r'(\d+)\s*GB\s*[\s/\-\(\)]*\s*(\d+)\s*(GB|TB)', n2, re.I)
    if m: return int(m.group(1)), int(m.group(2))*(1024 if m.group(3).upper()=='TB' else 1)
    m = re.search(r'(\d+)\s*(GB|TB)', n2, re.I)
    if m:
        v = int(m.group(1))*(1024 if m.group(2).upper()=='TB' else 1)
        return (np.nan, v) if v >= 32 else (v, np.nan)
    return np.nan, np.nan

def conn(n):
    l = n.lower()
    if '5g' in l: return '5G'
    if 'wifi' in l or 'wi-fi' in l: return 'Wifi'
    if '4g' in l: return '4G'
    return 'Unspecified'

GROUP = {'Apple':'Apple','Samsung':'Samsung','Masstel':'Vietnamese brands','Viettel':'Vietnamese brands','Vsmart':'Vietnamese brands',
         'Mobell':'Vietnamese brands','Bphone':'Vietnamese brands','Benco':'Vietnamese brands','Google':'Other international',
         'Nokia':'Other international','Sony':'Other international','Motorola':'Other international','Nothing':'Other international',
         'ASUS':'Other international','HTC':'Other international','Sharp':'Other international','Amazon':'Other international',
         'Acer':'Other international'}

df['brand'] = df.name.apply(brand_of)
df['brand_group'] = df.brand.map(GROUP).fillna('Chinese brands')
df['product_type'] = np.where(df.name.str.contains('Máy đọc sách', case=False), 'E-reader', np.where(df.category=='tablet', 'Tablet', 'Phone'))
r = df.name.apply(mem); df['ram_gb'] = [x[0] for x in r]; df['storage_gb'] = [x[1] for x in r]
er = df.product_type == 'E-reader'
fix = er & df.ram_gb.notna() & df.storage_gb.isna()
df.loc[fix, 'storage_gb'] = df.loc[fix, 'ram_gb']; df.loc[er, 'ram_gb'] = np.nan   # e-reader: single GB figure = storage
df['connectivity'] = df.name.apply(conn)
df['is_official'] = np.where(df.name.str.contains('Chính hãng', case=False), 'Official', 'Not stated')
df['has_bundle'] = np.where(df.name.str.contains('Kèm', case=False), 'Bundle', 'No bundle')
df['model_year'] = df.name.str.extract(r'\b(20[12]\d)\b')[0].astype(float)
df['has_discount'] = np.where(df.discount.notna(), 'Discount shown', 'No discount shown')
df['storage_tier'] = pd.cut(df.storage_gb, [0, 64, 128, 256, 1e9], labels=['<=64 GB', '128 GB', '256 GB', '>=512 GB']).astype(object).fillna('Unknown')
df['ram_tier'] = pd.cut(df.ram_gb, [0, 4, 8, 1e9], labels=['<=4 GB', '6-8 GB', '>=12 GB']).astype(object).fillna('Unknown')
df['availability'] = np.where(df.is_contact, 'Contact (no price)', 'Priced')

N = len(df); P = df[df.price_sale.notna()].copy()
RES['N'] = N; RES['n_priced'] = len(P); RES['n_contact'] = int(df.is_contact.sum())
RES['derived_cov'] = {c: int(df[c].notna().sum()) for c in ['ram_gb', 'storage_gb', 'model_year']}
RES['brand_counts'] = df.brand.value_counts().to_dict(); RES['n_brands'] = df.brand.nunique()
RES['group_counts'] = df.brand_group.value_counts().to_dict()
RES['type_counts'] = df.product_type.value_counts().to_dict()
RES['cat_counts'] = df.category.value_counts().to_dict()
RES['conn_counts'] = df.connectivity.value_counts().to_dict()
RES['priced_by_group'] = P.brand_group.value_counts().to_dict()
RES['priced_by_type'] = P.product_type.value_counts().to_dict()
RES['contact_by_cat'] = pd.crosstab(df.category, df.is_contact).to_dict('index')

# Fig 0: dataset overview (listings by brand, priced vs contact)
top = df.brand.value_counts().head(12).index
ov = pd.crosstab(df[df.brand.isin(top)].brand, df[df.brand.isin(top)].availability).loc[top]
ov.plot(kind='barh', stacked=True, figsize=(7, 4.2), color=['#BDBDBD', '#1F77B4'])
plt.gca().invert_yaxis(); plt.xlabel('Number of product listings'); plt.ylabel('Brand')
plt.title('Figure 1. Listings per brand (top 12): priced vs "Contact" (no price)'); plt.legend(title='Availability')
savefig('fig0_overview')
```

```python
attr_table = pd.DataFrame([
 ['category', 'Original', 'Nominal', 'Two labels (mobile/tablet) with no order; only equality is meaningful.', 'Grouping variable; dimension in the data cube.'],
 ['name', 'Original', 'Nominal', 'Free-text product title; identifies a product but has no order or arithmetic.', 'Source for brand, RAM, storage, connectivity extraction.'],
 ['url', 'Original', 'Nominal', 'Unique page address; pure identifier.', 'Duplicate check; excluded from modelling.'],
 ['is_contact', 'Original', 'Nominal (binary)', 'True/False flag for "price on request"; categories only.', 'Availability dimension; explains missing prices.'],
 ['price_sale', 'Original', 'Ratio', 'Money amount in VND with a true zero; ratios are meaningful (2x as expensive).', 'Descriptive statistics, distance, class target (price level).'],
 ['price_original', 'Original', 'Ratio', 'List price before promotion in VND; true zero.', 'Correlation with sale price; discount check.'],
 ['discount', 'Original', 'Ratio', 'Percentage off list price; 0% means no discount and 40% is twice 20%.', 'Variability, outliers, similarity attribute.'],
 ['brand', 'Derived', 'Nominal', 'Manufacturer extracted from name; no natural order.', 'Concept hierarchy, chi-square, information gain.'],
 ['ram_gb', 'Derived', 'Ratio', 'Memory in GB with true zero; 12 GB is 3x of 4 GB.', 'Correlation with price; ordinal tiers for classification.'],
 ['storage_gb', 'Derived', 'Ratio', 'Capacity in GB (TB converted to GB); true zero.', 'Distance, correlation, visualization encoding.'],
 ['connectivity', 'Derived', 'Nominal', '5G / 4G / Wifi / Unspecified network label; no order.', 'Chi-square and information gain.'],
 ['price_level', 'Derived', 'Ordinal', 'Low < Medium < High; order exists but gaps are unequal.', 'Class (target) variable for classification.'],
 ['storage_tier', 'Derived', 'Ordinal', '<=64 < 128 < 256 < >=512 GB; ordered categories.', 'Candidate predictor (information gain).'],
 ['model_year', 'Derived', 'Interval', 'Calendar year in the title (2019-2026); differences meaningful but year 0 is arbitrary, so no ratios.', 'Only 31 listings contain it; descriptive use only.'],
], columns=['Attribute', 'Source', 'Type', 'Why it belongs to this type', 'Possible use in later analysis'])
```

```python
def describe(s, name):
    s = s.dropna(); q1, med, q3 = s.quantile([.25, .5, .75]); iqr = q3 - q1
    lo, hi = q1 - 1.5*iqr, q3 + 1.5*iqr
    return {'Attribute': name, 'n': len(s), 'Mean': s.mean(), 'Median': med, 'Min': s.min(), 'Max': s.max(), 'Range': s.max()-s.min(),
            'Std': s.std(), 'CV (%)': s.std()/s.mean()*100, 'Q1': q1, 'Q3': q3, 'IQR': iqr, 'Skewness': stats.skew(s),
            'Lower fence': lo, 'Upper fence': hi, 'Outliers': int(((s < lo) | (s > hi)).sum())}
desc = pd.DataFrame([describe(P.price_sale/M, 'Sale price (M VND)'), describe(df.price_original/M, 'Original price (M VND)'),
                     describe(df.discount, 'Discount (%)'), describe(df.storage_gb, 'Storage (GB)'), describe(df.ram_gb, 'RAM (GB)')]).set_index('Attribute')
desc['Outlier %'] = desc.Outliers/desc.n*100
by_type = P.groupby('product_type').price_sale.agg(['count', 'mean', 'median', 'std', 'min', 'max']).assign(**{}).copy()
for c in ['mean', 'median', 'std', 'min', 'max']: by_type[c] = by_type[c]/M
by_group = P.groupby('brand_group').agg(n=('price_sale', 'size'), mean=('price_sale', lambda s: s.mean()/M), median=('price_sale', lambda s: s.median()/M),
                                        disc_mean=('discount', 'mean'))
RES['top_expensive'] = P.nlargest(3, 'price_sale')[['name', 'price_sale']].values.tolist()
RES['top_cheap'] = P.nsmallest(3, 'price_sale')[['name', 'price_sale']].values.tolist()
RES['top_disc'] = df.nlargest(3, 'discount')[['name', 'discount', 'price_sale']].values.tolist()
RES['share_price_gt_30m'] = float((P.price_sale > 30e6).mean()*100)
```

```python
# Fig 2 histogram
fig, ax = plt.subplots(1, 2, figsize=(9, 3.6))
x = P.price_sale/M
ax[0].hist(x, bins=25, color='#1F77B4', edgecolor='white'); ax[0].axvline(x.mean(), color='#D62728', ls='--', label=f'Mean = {x.mean():.1f}')
ax[0].axvline(x.median(), color='#2CA02C', ls='-', label=f'Median = {x.median():.1f}'); ax[0].legend()
ax[0].set_xlabel('Sale price (million VND)'); ax[0].set_ylabel('Number of products'); ax[0].set_title('(a) Linear scale')
ax[1].hist(np.log10(x), bins=25, color='#FF7F0E', edgecolor='white'); ax[1].set_xlabel('log10 of sale price (million VND)')
ax[1].set_ylabel('Number of products'); ax[1].set_title('(b) Log10 scale')
fig.suptitle('Figure 2. Histogram of sale price (n = %d priced products)' % len(x), y=1.03); savefig('fig1_hist')

# Fig 3 boxplots
order = P.groupby('brand_group').price_sale.median().sort_values().index
fig, ax = plt.subplots(1, 2, figsize=(9.5, 3.8))
sns.boxplot(data=P.assign(p=P.price_sale/M), y='brand_group', x='p', order=order, palette=PAL, ax=ax[0], fliersize=3)
ax[0].set_xlabel('Sale price (million VND)'); ax[0].set_ylabel('Brand group'); ax[0].set_title('(a) Sale price by brand group')
d = df.dropna(subset=['discount']); order2 = d.groupby('brand_group').discount.median().sort_values().index
sns.boxplot(data=d, y='brand_group', x='discount', order=order2, palette=PAL, ax=ax[1], fliersize=3)
ax[1].set_xlabel('Discount (%)'); ax[1].set_ylabel(''); ax[1].set_title('(b) Discount by brand group')
fig.suptitle('Figure 3. Boxplots of sale price and discount', y=1.03); savefig('fig2_box')

# Fig 4 quantile plot
s = np.sort(P.price_sale.values)/M; f = (np.arange(1, len(s)+1)-0.5)/len(s)
fig, ax = plt.subplots(figsize=(6.5, 4))
ax.plot(f, s, 'o', ms=3, color='#1F77B4')
for q, lab in [(.25, 'Q1'), (.5, 'Median'), (.75, 'Q3')]:
    v = np.quantile(s, q); ax.axvline(q, color='grey', ls=':'); ax.annotate(f'{lab}={v:.1f}', (q, v), textcoords='offset points', xytext=(6, 8), fontsize=8)
ax.set_xlabel('f-value (fraction of products with price <= y)'); ax.set_ylabel('Sale price (million VND)')
ax.set_title('Figure 4. Quantile plot of sale price'); savefig('fig3_quantile')
RES['quantile_pts'] = {str(q): float(np.quantile(s, q)) for q in [.1, .25, .5, .75, .9]}

# Fig 5 scatter
c2 = df.dropna(subset=['price_original', 'discount'])
fig, ax = plt.subplots(1, 2, figsize=(9.5, 3.8))
ax[0].scatter(c2.price_original/M, c2.price_sale/M, s=14, alpha=.7, color='#1F77B4'); lim = [0, c2.price_original.max()/M*1.03]
ax[0].plot(lim, lim, 'k--', lw=1, label='No discount (y = x)'); ax[0].legend()
ax[0].set_xlabel('Original price (million VND)'); ax[0].set_ylabel('Sale price (million VND)'); ax[0].set_title('(a) Sale vs original price')
ax[1].scatter(c2.price_sale/M, c2.discount, s=14, alpha=.7, color='#D62728'); ax[1].set_xlabel('Sale price (million VND)'); ax[1].set_ylabel('Discount (%)')
ax[1].set_title('(b) Discount vs sale price')
fig.suptitle('Figure 5. Scatter plots (n = %d products with discount)' % len(c2), y=1.03); savefig('fig4_scatter')

# Fig 6 multivariate
m5 = P.dropna(subset=['storage_gb']).copy(); rng = np.random.default_rng(42)
m5['xj'] = np.log2(m5.storage_gb) + rng.uniform(-.15, .15, len(m5))
mk = {'Phone': 'o', 'Tablet': 's', 'E-reader': '^'}
fig, ax = plt.subplots(figsize=(8, 4.8))
for (g, t), sub in m5.groupby(['brand_group', 'product_type']):
    ax.scatter(sub.xj, sub.price_sale/M, c=PAL[g], marker=mk[t], s=28, alpha=.75, edgecolors='white', linewidths=.4)
ax.set_yscale('log'); ax.set_yticks([1, 2, 5, 10, 20, 50]); ax.set_yticklabels(['1', '2', '5', '10', '20', '50'])
tk = sorted(m5.storage_gb.unique()); ax.set_xticks(np.log2(tk)); ax.set_xticklabels([f'{int(t)}' for t in tk])
ax.set_xlabel('Storage capacity (GB, log2 scale, jittered)'); ax.set_ylabel('Sale price (million VND, log scale)')
h1 = [plt.Line2D([], [], marker='o', ls='', color=c, label=g) for g, c in PAL.items()]
h2 = [plt.Line2D([], [], marker=m, ls='', color='k', label=t) for t, m in mk.items()]
l1 = ax.legend(handles=h1, title='Brand group (colour)', loc='upper left', fontsize=8); ax.add_artist(l1)
ax.legend(handles=h2, title='Product type (shape)', loc='lower right', fontsize=8)
ax.set_title('Figure 6. Price vs storage by brand group and product type (n = %d)' % len(m5)); savefig('fig5_multi')
RES['median_price_by_storage'] = (m5.groupby('storage_gb').price_sale.median()/M).round(2).to_dict()
RES['n_multi'] = len(m5)
```

```python
feat = ['price_sale', 'storage_gb', 'discount']
pop = df.dropna(subset=feat); mu, sd = pop[feat].mean(), pop[feat].std()
RES['sim_pop_n'] = len(pop); RES['sim_mu'] = mu.to_dict(); RES['sim_sd'] = sd.to_dict()
picks = ['iPhone 17 Pro 256GB | Chính hãng', 'iPhone 17 Pro Max 256GB | Chính hãng', 'iPhone 16 Plus 128GB | Chính hãng VN/A',
         'iPad Air 11 inch M3 5G 128GB 2025 | Chính hãng Apple Việt Nam', 'Samsung Galaxy S25 256GB', 'Samsung Galaxy Z Fold7 12GB 256GB',
         'Samsung Galaxy A07 5G 4GB 128GB', 'Samsung Galaxy A17 5G 8GB 128GB', 'Xiaomi Redmi Note 15 6GB 128GB',
         'Xiaomi Redmi Note 14 6GB 128GB', 'OPPO A3 6GB 128GB', 'Samsung Galaxy Tab A11 Wifi 4GB 64GB']
S = df[df.name.isin(picks)].set_index('name').loc[picks][feat].copy()
assert len(S) == 12 and S.notna().all().all()
short = ['iPhone 17 Pro 256', 'iPhone 17 Pro Max 256', 'iPhone 16 Plus 128', 'iPad Air 11 M3 128', 'Galaxy S25 256', 'Galaxy Z Fold7 256',
         'Galaxy A07 128', 'Galaxy A17 128', 'Redmi Note 15 128', 'Redmi Note 14 128', 'OPPO A3 128', 'Galaxy Tab A11 64']
S.index = short; Z = (S - mu)/sd
from scipy.spatial.distance import pdist, squareform
DE = pd.DataFrame(squareform(pdist(Z, 'euclidean')), index=short, columns=short)
DM = pd.DataFrame(squareform(pdist(Z, 'cityblock')), index=short, columns=short)
iu = np.triu_indices(12, 1)
pairs = pd.DataFrame({'A': np.array(short)[iu[0]], 'B': np.array(short)[iu[1]], 'Euclidean': DE.values[iu], 'Manhattan': DM.values[iu]})
pairs['rank_E'] = pairs.Euclidean.rank(); pairs['rank_M'] = pairs.Manhattan.rank()
near_E = pairs.nsmallest(3, 'Euclidean'); far_E = pairs.nlargest(3, 'Euclidean')
near_M = pairs.nsmallest(3, 'Manhattan'); far_M = pairs.nlargest(3, 'Manhattan')
RES['sim_overlap_near'] = len(set(map(tuple, near_E[['A', 'B']].values)) & set(map(tuple, near_M[['A', 'B']].values)))
RES['sim_overlap_far'] = len(set(map(tuple, far_E[['A', 'B']].values)) & set(map(tuple, far_M[['A', 'B']].values)))
RES['sim_spearman'] = float(stats.spearmanr(pairs.Euclidean, pairs.Manhattan)[0])
fig, ax = plt.subplots(figsize=(8, 6.4))
sns.heatmap(DE, annot=True, fmt='.2f', cmap='YlOrRd', cbar_kws={'label': 'Euclidean distance (z-scores)'}, ax=ax, annot_kws={'fontsize': 7})
ax.set_title('Figure 7. Euclidean distance matrix of 12 products (price, storage, discount; z-scored)'); plt.xticks(rotation=60, ha='right')
savefig('fig6_dist')
```

```python
P['price_level'] = pd.qcut(P.price_sale, 3, labels=['Low', 'Medium', 'High']).astype(str)
cuts = P.price_sale.quantile([0, 1/3, 2/3, 1]).values/M; RES['cuts'] = cuts.tolist()
RES['class_counts'] = P.price_level.value_counts().reindex(['Low', 'Medium', 'High']).to_dict()
RES['level_ranges'] = P.groupby('price_level').price_sale.agg(['min', 'max']).div(M).to_dict('index')

def entropy(s):
    p = s.value_counts(normalize=True); return float(-(p*np.log2(p)).sum())
def info_gain(data, attr, target='price_level'):
    H = entropy(data[target]); w = data[attr].value_counts(normalize=True)
    cond = sum(w[v]*entropy(data.loc[data[attr] == v, target]) for v in w.index)
    split = float(-(w*np.log2(w)).sum()); ig = H - cond
    return {'Attribute': attr, 'Levels': data[attr].nunique(), 'H(class|attr)': cond, 'Info gain': ig, 'Split info': split, 'Gain ratio': ig/split if split > 0 else 0}
H_class = entropy(P.price_level); RES['H_class'] = H_class
# alternative discretisation schemes (for comparison of class balance / entropy)
alt = {'Equal-frequency (terciles) - chosen': P.price_level,
       'Equal-width (3 bins of equal VND range)': pd.cut(P.price_sale, 3, labels=['Low', 'Medium', 'High']).astype(str),
       'Domain thresholds (<5M / 5-15M / >=15M)': pd.cut(P.price_sale, [0, 5e6, 15e6, 1e12], right=False, labels=['Low', 'Medium', 'High']).astype(str)}
alt_tab = pd.DataFrame([{'Scheme': k, **{l: int((v == l).sum()) for l in ['Low', 'Medium', 'High']}, 'Entropy (bits)': entropy(v)} for k, v in alt.items()])
cands = ['brand_group', 'product_type', 'connectivity', 'storage_tier', 'ram_tier', 'is_official', 'has_bundle', 'has_discount']
ig_tab = pd.DataFrame([info_gain(P, a) for a in cands]).sort_values('Info gain', ascending=False).reset_index(drop=True)
fig, ax = plt.subplots(figsize=(6.5, 3.6))
ax.barh(ig_tab.Attribute[::-1], ig_tab['Info gain'][::-1], color='#1F77B4'); ax.set_xlabel('Information gain (bits)'); ax.set_ylabel('Candidate attribute')
ax.set_title('Figure 8. Information gain for class = price level'); savefig('fig7_ig')

P['brand_group4'] = P.brand_group.replace({'Vietnamese brands': 'Other brands', 'Other international': 'Other brands'})
P['category_'] = P.category
chi_rows = []; chi_cont = {}
for a, lab in [('brand_group4', 'Brand group (4 levels)'), ('category_', 'Category (mobile/tablet)'), ('connectivity', 'Connectivity'), ('storage_tier', 'Storage tier')]:
    ct = pd.crosstab(P[a], P.price_level)[['Low', 'Medium', 'High']]; chi2, p, dof, exp = stats.chi2_contingency(ct)
    chi_rows.append({'Attribute': lab, 'Chi-square': chi2, 'dof': dof, 'p-value': p, "Cramer's V": np.sqrt(chi2/(ct.values.sum()*(min(ct.shape)-1))),
                     'Cells with E<5 (%)': (exp < 5).mean()*100, 'Min expected': exp.min()})
    chi_cont[lab] = ct
chi_tab = pd.DataFrame(chi_rows)
pairs_corr = [('price_sale', 'price_original'), ('price_sale', 'storage_gb'), ('price_sale', 'ram_gb'), ('ram_gb', 'storage_gb'), ('price_sale', 'discount')]
rows = []
for a, b in pairs_corr:
    d_ = df.dropna(subset=[a, b]); pr, pp = stats.pearsonr(d_[a], d_[b]); sr, sp = stats.spearmanr(d_[a], d_[b])
    rows.append({'Pair': f'{a} vs {b}', 'n': len(d_), 'Pearson r': pr, 'p (Pearson)': pp, 'Spearman rho': sr, 'p (Spearman)': sp})
corr_tab = pd.DataFrame(rows)
cm = df[['price_sale', 'price_original', 'discount', 'storage_gb', 'ram_gb']].corr(method='spearman')
fig, ax = plt.subplots(figsize=(5.6, 4.4)); sns.heatmap(cm, annot=True, fmt='.2f', cmap='RdBu_r', vmin=-1, vmax=1, ax=ax, cbar_kws={'label': 'Spearman rho'})
ax.set_title('Figure 9. Spearman correlation matrix (pairwise complete)'); plt.xticks(rotation=35, ha='right'); savefig('fig8_corr')
```

```python
def pband(v):
    if pd.isna(v): return 'On request'
    for hi, lab in [(3e6, '<3M'), (5e6, '3-5M'), (10e6, '5-10M'), (15e6, '10-15M'), (20e6, '15-20M'), (30e6, '20-30M'), (50e6, '30-50M')]:
        if v < hi: return lab
    return '>=50M'
TIER = {'<3M': 'Budget', '3-5M': 'Budget', '5-10M': 'Mid-range', '10-15M': 'Mid-range', '15-20M': 'Premium', '20-30M': 'Premium', '30-50M': 'Flagship', '>=50M': 'Flagship', 'On request': 'On request'}
df['price_band'] = df.price_sale.apply(pband); df['price_tier'] = df.price_band.map(TIER)
TORD = ['Budget', 'Mid-range', 'Premium', 'Flagship', 'On request']
hier1 = pd.DataFrame({'Level': ['Product (name)', 'Brand', 'Brand group'], 'Distinct values': [df.name.nunique(), df.brand.nunique(), df.brand_group.nunique()]})
hier2 = pd.DataFrame({'Level': ['Exact price (priced products)', 'Price band', 'Price tier'], 'Distinct values': [P.price_sale.nunique(), df[df.price_tier != 'On request'].price_band.nunique(), df[df.price_tier != 'On request'].price_tier.nunique()]})
tier_tab = df[df.price_tier != 'On request'].groupby(['price_tier', 'price_band']).size().rename('Products').reset_index()
# data cube
cube3 = df.groupby(['brand_group', 'product_type', 'price_tier']).agg(Products=('name', 'size'), Avg_price_M=('price_sale', lambda s: s.mean()/M),
                                                                       Avg_discount=('discount', 'mean')).reset_index()
cube2 = pd.crosstab(df.brand_group, df.price_tier)[TORD]; cube2['Total'] = cube2.sum(axis=1)
cube_roll = df.groupby('brand_group').agg(Products=('name', 'size'), Priced=('price_sale', 'count'), Avg_price_M=('price_sale', lambda s: s.mean()/M),
                                          Median_price_M=('price_sale', lambda s: s.median()/M), Avg_discount=('discount', 'mean')).sort_values('Products', ascending=False)
slice_phone = df[df.product_type == 'Phone'].groupby(['brand_group', 'price_tier']).agg(Products=('name', 'size'), Avg_price_M=('price_sale', lambda s: s.mean()/M)).reset_index()
drill_apple = df[df.brand_group == 'Apple'].groupby(['product_type', 'price_tier']).agg(Products=('name', 'size'), Avg_price_M=('price_sale', lambda s: s.mean()/M), Avg_discount=('discount', 'mean')).reset_index()
# Attribute-oriented induction
aoi0 = df[['name', 'brand', 'product_type', 'price_sale', 'storage_gb', 'connectivity']].copy()
aoi = pd.DataFrame({'brand_group': df.brand_group, 'product_type': df.product_type, 'price_tier': df.price_tier,
                    'storage': df.storage_tier.replace({'<=64 GB': 'Small (<=128 GB)', '128 GB': 'Small (<=128 GB)', '256 GB': 'Large (>=256 GB)', '>=512 GB': 'Large (>=256 GB)'}),
                    'network': df.connectivity.replace({'5G': 'Cellular', '4G': 'Cellular', 'Unspecified': 'Unspecified'})})
steps = [('Original relation (name, brand, type, price, storage, connectivity)', len(aoi0.drop_duplicates(subset=['brand', 'product_type', 'price_sale', 'storage_gb', 'connectivity']))),
         ('Remove name (no generalization operator, ~1,018 distinct values)', len(aoi0.drop_duplicates(subset=['brand', 'product_type', 'price_sale', 'storage_gb', 'connectivity']))),
         ('Generalize brand -> brand group, price -> price tier, storage -> storage class, connectivity -> network', len(aoi.drop_duplicates()))]
aoi_rel = aoi.groupby(list(aoi.columns)).size().rename('Count').reset_index().sort_values('Count', ascending=False)
aoi_rel['t-weight (%)'] = aoi_rel.Count/N*100
RES['aoi_steps'] = steps; RES['aoi_nrows'] = len(aoi_rel)
# second round: merge phone/tablet/e-reader rows into 'Phone' vs 'Tablet/E-reader' and drop network to reach <= 12 tuples
aoi2 = aoi.drop(columns=['network']).copy(); aoi2['product_type'] = aoi2.product_type.replace({'Tablet': 'Tablet/E-reader', 'E-reader': 'Tablet/E-reader'})
aoi2_rel = aoi2.groupby(list(aoi2.columns)).size().rename('Count').reset_index().sort_values('Count', ascending=False)
aoi2_rel['t-weight (%)'] = aoi2_rel.Count/N*100
RES['aoi2_nrows'] = len(aoi2_rel)
aoi3 = aoi2[['brand_group', 'product_type', 'price_tier']]; aoi3_rel = aoi3.groupby(list(aoi3.columns)).size().rename('Count').reset_index().sort_values('Count', ascending=False)
aoi3_rel['t-weight (%)'] = aoi3_rel.Count/N*100
RES['aoi3_nrows'] = len(aoi3_rel)


# stage 4: climb the hierarchies again until the generalized relation has <= 18 tuples
aoi4 = pd.DataFrame({'brand_group': aoi2.brand_group.replace({'Chinese brands': 'Other brands', 'Vietnamese brands': 'Other brands', 'Other international': 'Other brands'}),
                     'product_type': aoi2.product_type,
                     'price_level': aoi2.price_tier.replace({'Budget': 'Mass-market (<15M)', 'Mid-range': 'Mass-market (<15M)', 'Premium': 'High-end (>=15M)', 'Flagship': 'High-end (>=15M)'})})
aoi4_rel = aoi4.groupby(list(aoi4.columns)).size().rename('Count').reset_index().sort_values('Count', ascending=False)
aoi4_rel['t-weight (%)'] = aoi4_rel.Count/N*100
RES['aoi4_nrows'] = len(aoi4_rel)
```
