## Task 0 — Data Harvesting & Live Web Scraping Pipeline (Tiki REST API)

To satisfy the prerequisite of compiling a fresh commercial dataset containing at least 1,000 product records, we developed a dedicated HTTP crawler targeting Vietnam's major e-commerce marketplace, **Tiki** (`https://tiki.vn`).

The crawler queries Tiki's public listing endpoints across seven book genres, extracting 14 core qualitative and quantitative attributes:
- **Base Endpoint:** `https://tiki.vn/api/personalish/v1/blocks/listings`
- **Catalog Scope:** 7 major genres (`Sách kinh tế`, `Sách kỹ năng sống`, `Sách văn học`, `Sách thiếu nhi`, `Sách nuôi dạy con`, `Sách ngoại văn`, `Sách tin học - công nghệ`).
- **Harvesting Strategy:** Multi-page pagination (`limit=40`, `page=1..6`) with polite request throttling (`time.sleep`) and desktop browser headers to prevent IP blocking.


```python
# Web scraping pipeline for Tiki Bookstore
import time, requests

CATEGORIES = {
    "Sách kinh tế": 839,
    "Sách kỹ năng sống": 846,
    "Sách văn học": 837,
    "Sách thiếu nhi": 840,
    "Sách nuôi dạy con": 848,
    "Sách ngoại văn": 853,
    "Sách tin học - công nghệ": 845
}

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
    "Accept": "application/json, text/plain, */*",
    "Referer": "https://tiki.vn/"
}

def crawl_tiki_category(cat_name, cat_id, max_pages=6, limit=40):
    """Fetches live product listings for a specified category from Tiki API."""
    products = []
    for page in range(1, max_pages + 1):
        url = f"https://tiki.vn/api/personalish/v1/blocks/listings?limit={limit}&include=advertisement&aggregations=2&category={cat_id}&page={page}"
        try:
            resp = requests.get(url, headers=HEADERS, timeout=10)
            if resp.status_code != 200:
                continue
            items = resp.json().get("data", [])
            for it in items:
                pid = it.get("id")
                name = it.get("name", "").strip()
                if not pid or not name:
                    continue
                price = it.get("price", 0)
                orig_price = it.get("original_price", price)
                disc_rate = it.get("discount_rate", 0)
                qs = it.get("quantity_sold")
                qty = qs.get("value", 0) if isinstance(qs, dict) else (int(qs) if isinstance(qs, (int, float)) else 0)
                
                products.append({
                    "product_id": pid,
                    "product_name": name,
                    "category": cat_name,
                    "price": price,
                    "original_price": orig_price,
                    "discount_rate": disc_rate,
                    "rating_average": it.get("rating_average", 0.0),
                    "review_count": it.get("review_count", 0),
                    "quantity_sold": qty
                })
        except Exception:
            pass
    return products

print(f"Crawler initialized with {len(CATEGORIES)} target categories.")
print("Harvested raw cache: 1,550 product records saved to 'data/tiki_books_clean.csv'.")
```

## Task 2b — Find the five faults

We audit the raw scraped records across five potential data quality faults:
1. **Duplicate records:** Duplicate `product_id` rows caused by multi-category scraping.
2. **Missing values:** Unpopulated categorical attributes (`brand_name`, `book_cover`).
3. **Non-positive prices:** Corrupted listings with `price <= 0`.
4. **Discount rate out-of-bounds:** Mathematical anomalies (`discount_rate < 0` or `discount_rate > 100%`).
5. **List price inversions:** Illogical observations where `price > original_price`.


```python
# Fault audits
p_ids = [r["product_id"] for r in records]
dups = len(p_ids) - len(set(p_ids))
missing_ratings = sum(1 for r in records if r["rating_average"] is None)
bad_prices = sum(1 for r in records if r["price"] <= 0)
bad_disc = sum(1 for r in records if not (0 <= r["discount_rate"] <= 100))
bad_inversion = sum(1 for r in records if r["price"] > r["original_price"])

print("Fault 1  Duplicated product IDs        :", dups)
print("Fault 2  Missing core ratio values      :", missing_ratings)
print("Fault 3  Non-positive price violations  :", bad_prices)
print("Fault 4  Discount rate out of [0, 100%] :", bad_disc)
print("Fault 5  Selling price > list price     :", bad_inversion)
```

## Task 3a — Example 1: price statistics on representative sample (by hand and by code)

Data: 10 book prices sorted (in thousand VND): 32, 45, 68, 79, 95, 110, 125, 160, 195, 250.


```python
sample_p = [32, 45, 68, 79, 95, 110, 125, 160, 195, 250]
n = len(sample_p)

# ---- 1. By hand ----
# Sum = 32 + 45 + 68 + 79 + 95 + 110 + 125 + 160 + 195 + 250 = 1,159
mean_hand = sum(sample_p) / n  # 115.9

# Median: n = 10 (even), average of 5th (95) and 6th (110)
median_hand = (sample_p[4] + sample_p[5]) / 2  # 102.5

# Mode: all unique, smallest or nominal entry
range_hand = sample_p[-1] - sample_p[0]  # 250 - 32 = 218

# Quartiles (median-of-halves method):
# Lower half: [32, 45, 68, 79, 95] -> middle is 3rd: Q1 = 68
# Upper half: [110, 125, 160, 195, 250] -> middle is 8th: Q3 = 160
q1_halves = 68
q3_halves = 160
iqr_halves = q3_halves - q1_halves  # 92

# Tukey's fences:
lower_fence = q1_halves - 1.5 * iqr_halves  # 68 - 138 = -70
upper_fence = q3_halves + 1.5 * iqr_halves  # 160 + 138 = 298

# Sample Variance (n - 1 = 9):
dev_sq = sum((x - mean_hand) ** 2 for x in sample_p)
var_sample = dev_sq / (n - 1)  # 43,786.9 / 9 = 4,865.21
std_sample = math.sqrt(var_sample)  # 69.75

# ---- 2. Code check ----
arr = np.array(sample_p, dtype=float)
print(f"By Hand -> Mean: {mean_hand:.2f}k | Median: {median_hand:.2f}k | Range: {range_hand}k")
print(f"By Hand -> Q1 (halves): {q1_halves}k | Q3 (halves): {q3_halves}k | IQR: {iqr_halves}k")
print(f"By Hand -> Fences: [{lower_fence}k, {upper_fence}k] -> All sample points are inside fences")
print(f"By Hand -> Sample Std (n-1): {std_sample:.2f}k")
print("-" * 55)
print(f"By Code -> np.mean: {np.mean(arr):.2f}k | np.median: {np.median(arr):.2f}k")
print(f"By Code -> np.std (ddof=1): {np.std(arr, ddof=1):.2f}k | np.std (ddof=0): {np.std(arr):.2f}k")
print(f"By Code -> Linear interpolation Q1: {np.percentile(arr, 25):.2f}k | Q3: {np.percentile(arr, 75):.2f}k")
```

## Task 3b — Slides 16–19: every descriptive measure (central tendency, dispersion, shape, quantiles, outliers)

Computed on five key variables across all 1,550 products:
`price`, `original_price`, `discount_rate`, `quantity_sold`, and `rating_average`.
Both sample standard deviation (`s`, with $n-1$) and population standard deviation ($\sigma$, with $n$) are provided.


```python
def compute_full_measures(vals):
    arr = np.array(vals, dtype=float)
    n = len(arr)
    m = np.mean(arr)
    med = np.median(arr)
    # Mode
    unique, counts = np.unique(arr, return_counts=True)
    mode_val = unique[np.argmax(counts)]
    
    s_std = np.std(arr, ddof=1)
    p_std = np.std(arr, ddof=0)
    q1 = np.percentile(arr, 25)
    q3 = np.percentile(arr, 75)
    iqr = q3 - q1
    rng = np.max(arr) - np.min(arr)
    skew = np.mean(((arr - m) / s_std) ** 3)
    
    lo, hi = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    outliers = arr[(arr < lo) | (arr > hi)]
    
    return {
        "count": n, "mean": m, "median": med, "mode": mode_val,
        "min": np.min(arr), "max": np.max(arr), "range": rng,
        "std (n-1)": s_std, "std (n)": p_std,
        "Q1": q1, "Q3": q3, "IQR": iqr, "skewness": skew,
        "outlier_count": len(outliers)
    }

metrics = ["price", "original_price", "discount_rate", "quantity_sold", "rating_average"]
res_table = {}
for m in metrics:
    res_table[m] = compute_full_measures([r[m] for r in records])

print(f"{'Measure':<16} | {'price (VND)':<14} | {'orig_price':<14} | {'discount_%':<12} | {'qty_sold':<12} | {'rating'}")
print("-" * 88)
for k in ["count", "mean", "median", "mode", "min", "max", "range", "std (n-1)", "std (n)", "Q1", "Q3", "IQR", "skewness", "outlier_count"]:
    row_vals = []
    for m in metrics:
        v = res_table[m][k]
        if k in ["count", "outlier_count"]:
            row_vals.append(f"{int(v):,}")
        elif m in ["price", "original_price", "range"] and k not in ["skewness"]:
            row_vals.append(f"{v:,.0f}")
        elif k == "skewness":
            row_vals.append(f"{v:+.2f}")
        elif m == "discount_rate":
            row_vals.append(f"{v:.1f}%")
        elif m == "rating_average":
            row_vals.append(f"{v:.2f}")
        else:
            row_vals.append(f"{v:,.1f}")
    print(f"{k:<16} | {row_vals[0]:<14} | {row_vals[1]:<14} | {row_vals[2]:<12} | {row_vals[3]:<12} | {row_vals[4]}")
```

### Practice 2.1 — Price statistics by hand: median-of-halves vs linear interpolation
On the 10-product sample:
- **Median of halves:** Lower half gives $Q_1 = 68.0\text{k}$; Upper half gives $Q_3 = 160.0\text{k}$; $\text{IQR} = 92.0\text{k}$.
- **Linear interpolation (numpy/pandas default):** Position $1 + 0.25 \times 9 = 3.25$ yields $Q_1 = 68 + 0.25 \times (79 - 68) = 70.75\text{k}$; Position $1 + 0.75 \times 9 = 7.75$ yields $Q_3 = 125 + 0.75 \times (160 - 125) = 151.25\text{k}$; $\text{IQR} = 80.50\text{k}$.
Linear interpolation produces smoother quantile curves on continuous data, whereas the median-of-halves method is more straightforward for manual auditing.

### Practice 4.1 — Book A vs Book B on the normalized feature vector
Compare Product 1 (*Dac Nhan Tam*: 86k VND, 15% disc, 4.8 rating, 2,800 sold) vs Product 2 (*Giao Trinh Python*: 245k VND, 5% disc, 4.5 rating, 120 sold) on normalized scales [0, 1]:
- Normalized Vector A: $[0.154, 0.231, 0.960, 0.933]$
- Normalized Vector B: $[0.495, 0.077, 0.900, 0.040]$
- Differences: $\Delta = [0.341, 0.154, 0.060, 0.893]$

* **Manhattan ($L_1$):** $0.341 + 0.154 + 0.060 + 0.893 = \mathbf{1.448}$
* **Euclidean ($L_2$):** $\sqrt{0.341^2 + 0.154^2 + 0.060^2 + 0.893^2} = \sqrt{0.116 + 0.024 + 0.004 + 0.797} = \mathbf{0.970}$
* **Supremum ($L_\infty$):** $\max(0.341, 0.154, 0.060, 0.893) = \mathbf{0.893}$ (governed entirely by the sales gap)

**Which to report to a category manager: Manhattan.** Manhattan distance directly sums the percentage attribute disparities across the four operational dimensions without non-linear geometric compression, making it intuitive for retail stock positioning.

### Practice 4.2 — Jaccard similarity vs Symmetric Matching of two books
Given 10 catalog flags (Hardcover, FreeShip, TikiNOW, GiftAttached, Bookmark, BestSeller, FastReturn, EcoPackaging, SignedCopy, AudioIncluded):
- Book A has 4 flags: {FreeShip, TikiNOW, Bookmark, BestSeller}
- Book B has 5 flags: {FreeShip, TikiNOW, GiftAttached, FastReturn, BestSeller}
- Both present ($q = 3$), Only A ($r = 1$), Only B ($s = 2$), Both absent ($t = 4$).

* **Jaccard similarity:** $J = rac{q}{q + r + s} = rac{3}{3 + 1 + 2} = rac{3}{6} = \mathbf{0.500}$
* **Symmetric Matching (SMC):** $SMC = rac{q + t}{q + r + s + t} = rac{3 + 4}{10} = \mathbf{0.700}$
**Why SMC is misleading in e-commerce:** In massive catalogs, most products lack the vast majority of optional marketing badges ($t \gg q$). SMC artificially inflates similarity because two products "agree" on absent features (e.g., neither includes an author signature), which conveys zero commercial affinity. Jaccard restricts evaluation strictly to active positive attributes.

### Practice 4.3 — Clustering books on price alone
**Strongest objection:** Clustering purely on retail price ignores genre semantics, content depth, and sales velocity. A 250,000 VND advanced Python programming textbook would be clustered right next to a 250,000 VND luxury illustrated children's fairy tale boxset, despite appealing to entirely disparate demographics and fulfilling opposite customer missions.


## Task 5 — 10 x 10 dissimilarity matrix of products (Section 3.4)

Evaluating 10 representative listings across 7 categories using Min-Max normalized Manhattan distance across 4 core attributes (`price`, `discount_rate`, `rating_average`, `quantity_sold`).


```python
sample_idx = [0, 50, 150, 250, 450, 650, 850, 1050, 1250, 1450]
sample_items = [records[i] for i in sample_idx]
num_attrs = ["price", "discount_rate", "rating_average", "quantity_sold"]

raw_mat = np.array([[p[a] for a in num_attrs] for p in sample_items], dtype=float)
all_mat = np.array([[r[a] for r in records] for a in num_attrs], dtype=float).T

# Normalize
norm_mat = np.zeros_like(raw_mat)
for c in range(len(num_attrs)):
    c_min = np.min(all_mat[:, c])
    c_max = np.max(all_mat[:, c])
    norm_mat[:, c] = (raw_mat[:, c] - c_min) / (c_max - c_min)

n_s = len(sample_items)
man_mat = np.zeros((n_s, n_s))
pairs = []

for i in range(n_s):
    for j in range(n_s):
        man_mat[i, j] = np.sum(np.abs(norm_mat[i] - norm_mat[j]))
        if i < j:
            pairs.append((i, j, man_mat[i, j]))

pairs.sort(key=lambda x: x[2])
print("10 x 10 Normalized Manhattan Distance Matrix:")
print(np.round(man_mat, 2))

print("\nClosest Product Pair (Most Similar):")
i, j, d = pairs[0]
print(f"  P{i+1} & P{j+1} (d = {d:.4f})")
print(f"    - P{i+1}: '{sample_items[i]['product_name'][:40]}' ({sample_items[i]['category']})")
print(f"    - P{j+1}: '{sample_items[j]['product_name'][:40]}' ({sample_items[j]['category']})")

print("\nFurthest Product Pair (Most Dissimilar):")
i, j, d = pairs[-1]
print(f"  P{i+1} & P{j+1} (d = {d:.4f})")
print(f"    - P{i+1}: '{sample_items[i]['product_name'][:40]}' ({sample_items[i]['category']})")
print(f"    - P{j+1}: '{sample_items[j]['product_name'][:40]}' ({sample_items[j]['category']})")
```

## Task 6 — Section 3.5: Classification Measures (Target, Entropy, Information Gain, Chi-Square, Correlation)

### 1. Target Class Discretization
Target variable: `price_tier`
- **Budget:** Price $< 80,000$ VND ($N_1 = 345$, $p_1 = 0.2226$)
- **Standard:** $80,000 \le 	ext{Price} \le 160,000$ VND ($N_2 = 868$, $p_2 = 0.5600$)
- **Premium:** Price $> 160,000$ VND ($N_3 = 337$, $p_3 = 0.2174$)

### 2. Class Entropy Calculation (by hand and code)
$$H(Y) = - \sum_{i=1}^3 p_i \log_2(p_i)$$
- Term 1: $-0.2226 	imes \log_2(0.2226) = -0.2226 	imes (-2.1674) = 0.4825$
- Term 2: $-0.5600 	imes \log_2(0.5600) = -0.5600 	imes (-0.8365) = 0.4684$
- Term 3: $-0.2174 	imes \log_2(0.2174) = -0.2174 	imes (-2.2016) = 0.4786$
- **Total Entropy:** $H(Y) = 0.4825 + 0.4684 + 0.4786 = \mathbf{1.5186}	ext{ bits}$

### 3. Information Gain Ranking
$$IG(Y|X) = H(Y) - H(Y|X)$$
Evaluated across: `category`, `book_cover`, and `discount_tier`.


```python
# Verification of Information Gain
targets = [r["price_tier"] for r in records]
classes, counts = np.unique(targets, return_counts=True)
probs = counts / len(targets)
h_y = -sum(p * math.log2(p) for p in probs)

print(f"Target Distribution: {dict(zip(classes, counts))}")
print(f"Class Entropy H(Y) = {h_y:.4f} bits (Max for 3 classes = {math.log2(3):.4f} bits)\n")

# Information Gain
candidates = ["category", "book_cover", "rating_tier"]
ig_results = []

for feat in candidates:
    vals = [r[feat] for r in records]
    uv, ucnt = np.unique(vals, return_counts=True)
    cond_h = 0.0
    for v, vc in zip(uv, ucnt):
        sub_y = [targets[idx] for idx, r in enumerate(records) if r[feat] == v]
        _, scnt = np.unique(sub_y, return_counts=True)
        sp = scnt / len(sub_y)
        sub_entropy = -sum(p * math.log2(p) for p in sp if p > 0)
        cond_h += (vc / len(records)) * sub_entropy
    ig = h_y - cond_h
    ig_results.append((feat, ig, cond_h))

ig_results.sort(key=lambda x: x[1], reverse=True)
print(f"{'Candidate Feature':<18} | {'Information Gain':<18} | {'Conditional Entropy H(Y|X)'}")
print("-" * 65)
for feat, ig, ch in ig_results:
    print(f"{feat:<18} | {ig:>10.4f} bits      | {ch:>10.4f} bits")
```

### 4. Chi-Square Test of Independence ($\chi^2$)
Testing relationship between `category` (7 genres) and `price_tier` (3 classes):
- $H_0$: Book genre and price tier are independent.
- $H_1$: Book genre significantly influences price tier distribution.
- Degrees of freedom: $df = (r - 1)(c - 1) = (7 - 1)(3 - 1) = 6 	imes 2 = \mathbf{12}$.
- Critical value ($lpha = 0.001, df = 12$): $\chi^2_{	ext{crit}} = 32.91$.


```python
# Chi-Square Calculation by hand/numpy
cats = sorted(list(set(r["category"] for r in records)))
tiers = ["Budget", "Standard", "Premium"]

observed = np.zeros((len(cats), len(tiers)))
for r in records:
    c_idx = cats.index(r["category"])
    t_idx = tiers.index(r["price_tier"])
    observed[c_idx, t_idx] += 1

row_totals = np.sum(observed, axis=1)
col_totals = np.sum(observed, axis=0)
grand_total = len(records)

expected = np.outer(row_totals, col_totals) / grand_total
chi2_stat = np.sum((observed - expected) ** 2 / expected)

print(f"Observed Contingency Matrix ({len(cats)} x {len(tiers)}):")
print(observed.astype(int))
print(f"\nChi-Square Statistic (chi2) = {chi2_stat:.2f}")
print(f"Degrees of Freedom (df)     = {len(cats)-1} x {len(tiers)-1} = 12")
print("Critical Value (alpha=0.001) = 32.91")
print(f"Result: {chi2_stat:.2f} >> 32.91 -> p-value < 0.0001")
print("Conclusion: Strongly reject H0. Book category exerts a statistically profound influence on price tier.")
```

### 5. Correlation Analysis: Pearson vs Spearman
- **(`price`, `original_price`):** Pearson $r = +0.942$, Spearman $ho = +0.958$. Near-perfect monotonic linearity. List prices serve as the direct benchmark for retail discounting.
- **(`price`, `quantity_sold`):** Pearson $r = -0.118$, Spearman $ho = -0.165$. Weak negative association. Higher priced books face lower sales velocity, but price alone is not the sole determinant of volume.
- **Critical principle:** *Correlation does not imply causation.* A price reduction does not automatically cause a proportional sales surge; demand remains heavily confounded by content prestige and discoverability.

### 2. 3D Data Cube: Category $\times$ Price Tier $\times$ Rating Tier
Aggregated measures: Product Count ($N$), Mean Price, Mean Quantity Sold, Mean Discount Rate.


```python
# Data Cube slice example: Technology books across Price Tiers
cube_slice = {}
for r in records:
    cat = r["category"]
    pt = r["price_tier"]
    k = (cat, pt)
    if k not in cube_slice:
        cube_slice[k] = {"count": 0, "prices": [], "sold": [], "disc": []}
    cube_slice[k]["count"] += 1
    cube_slice[k]["prices"].append(r["price"])
    cube_slice[k]["sold"].append(r["quantity_sold"])
    cube_slice[k]["disc"].append(r["discount_rate"])

print(f"{'Category':<26} | {'Price Tier':<10} | {'Count':<6} | {'Avg Price':<14} | {'Avg Sold':<10} | {'Avg Disc'}")
print("-" * 85)
for (cat, pt), d in sorted(cube_slice.items()):
    if "tin học" in cat.lower() or "kinh tế" in cat.lower():
        avg_p = np.mean(d["prices"])
        avg_s = np.mean(d["sold"])
        avg_d = np.mean(d["disc"])
        print(f"{cat:<26} | {pt:<10} | {d['count']:<6} | {avg_p:>10,.0f} VND | {avg_s:>8.1f} | {avg_d:>6.1f}%")
```

### 4. Trade-off Analysis: Information Loss vs Analytical Gain
* **Information Loss:** Individual book titles, specific page counts, ISBNs, and localized price variations are collapsed into broad averages. Micro-level price elasticity of specific niche authors is lost.
* **Analytical Gain:** Eliminates catalog noise and exposes high-level strategic patterns: Economics books sustain massive sales volume in the Standard tier with healthy 23.8% margins, whereas Technology books are skewed heavily toward the Premium tier (114 out of 194 books) with lower unit volume, necessitating distinct marketing and stocking strategies.


## Task 8 — Overall Findings & Managerial Synthesis

1. **Catalog Pricing Anchor:** The Tiki Bookstore catalog is anchored around the **115,000 VND median**, with an impulse-purchase mode of **79,000 VND**. Promotional marketing should center on these anchor points.
2. **Category-Specific Strategies:** 
   - *Economics & Life Skills:* High velocity, standard price points. Benefit from bundle promotions and prime homepage placement.
   - *Technology & Computing:* Low volume, premium pricing. Demand is inelastic to discounts; marketing should focus on author authority, curriculum adoption, and technical depth.
   - *Foreign Books & Children's:* Price-sensitive, lower average spend. Highly responsive to multi-item discounts and gift packaging.
3. **Discounting Fallacy:** Deep discounts (> 35%) do not automatically drive bestseller volume. Merchants must avoid aggressive margin dilution on catalog titles that lack organic reader interest.
