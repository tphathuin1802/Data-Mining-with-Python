# SAIGON BEANS DATA ANALYSIS

```python
import pandas as pd
```

```python
stores = pd.read_csv("C:/Dev/saigon_beans_stephen/data/saigon_bean_stores.csv")
products = pd.read_csv("C:/Dev/saigon_beans_stephen/data/saigon_bean_products.csv")
```

```python
def classify_columns(df, numeric_discrete_cols=None, id_cols=None):
    numeric_discrete_cols = numeric_discrete_cols or []
    id_cols = id_cols or []
    result = []
    for col in df.columns:
        if col in id_cols:
            attr_type = "Nominal (identifier)"
            disc_cont = "Discrete (but a label, not a quantity)"
        elif pd.api.types.is_numeric_dtype(df[col]):
            attr_type = "Numeric (ratio scale)"
            disc_cont = "Discrete" if col in numeric_discrete_cols else "Continuous"
        else:
            attr_type = "Nominal"
            disc_cont = "-"
        result.append({"column": col, "attribute_type": attr_type, "discrete_or_continuous": disc_cont})
    return pd.DataFrame(result)
```

```python

stores_class = classify_columns(stores, numeric_discrete_cols=["seats"], id_cols=["store_id"])
products_class = classify_columns(products, id_cols=["product_id"])
```
